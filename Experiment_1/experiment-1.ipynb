```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("OpenCV Version:", cv2.__version__)
print("NumPy Version:", np.__version__)
```

```python
# Read Image
img = cv2.imread("Lion.jpg")

if img is None:
    print("Image not found!")
else:
    print("Image loaded successfully.")
    print("Shape :", img.shape)
    print("Data Type :", img.dtype)
```

```python
cv2.imshow("Lion Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(9,6))
plt.imshow(rgb)
plt.title("Lion Image (RGB)")
plt.axis("off")
plt.show()
```

```python
height, width, channels = img.shape

print("Height :", height)
print("Width :", width)
print("Channels :", channels)
```

```python
print("Pixel Value at (150,250):")
print(img[150,250])
```

```python
modified = img.copy()

modified[150,250] = [255,0,0]

cv2.imshow("Modified Pixel", modified)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
roi = img[100:350,200:500]

cv2.imshow("Region of Interest", roi)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale Image", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
resized = cv2.resize(img, (500,350))

cv2.imshow("Resized Image", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
horizontal = cv2.flip(img,1)
vertical = cv2.flip(img,0)

cv2.imshow("Horizontal Flip", horizontal)
cv2.imshow("Vertical Flip", vertical)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rotated = cv2.rotate(img, cv2.ROTATE_180)

cv2.imshow("Rotated Image", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rectangle = img.copy()

cv2.rectangle(
    rectangle,
    (120,80),
    (500,320),
    (255,0,0),
    2
)

cv2.imshow("Rectangle", rectangle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
text = img.copy()

cv2.putText(
    text,
    "Experiment 1",
    (80,60),
    cv2.FONT_HERSHEY_SIMPLEX,
    1.2,
    (0,255,255),
    3
)

cv2.imshow("Text on Image", text)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
